```python
import numpy as np

# 1) Générateur de nombres aléatoires
#    La "graine" 42 permet d'obtenir EXACTEMENT les mêmes nombres à chaque exécution
rng = np.random.default_rng(42)

N = 5000  # nombre de lignes (d'exemples) à fabriquer

# 2) Tirer au hasard une valeur réaliste pour chaque capteur
temperature  = rng.uniform(5, 45, N)    # entre 5 et 45 °C
humidite_air = rng.uniform(20, 95, N)   # entre 20 et 95 %
humidite_sol = rng.uniform(0, 100, N)   # entre 0 et 100 %
ph           = rng.uniform(4, 9, N)     # entre 4 et 9
luminosite   = rng.uniform(0, 100, N)   # entre 0 et 100 %

# 3) Ranger le tout dans un tableau pandas (un "DataFrame")
df = pd.DataFrame({
    "temperature":  temperature.round(1),
    "humidite_air": humidite_air.round(1),
    "humidite_sol": humidite_sol.round(0).astype(int),
    "ph":           ph.round(1),
    "luminosite":   luminosite.round(0).astype(int),
})

print("Taille du tableau :", df.shape)   # (nombre de lignes, nombre de colonnes)
df.head(10)                              # afficher les 10 premières lignes
```
